```python
import pandas as pd
```

```python
data={'gene':["gene1","gene2","gene3","gene4","gene5","gene6"],
      'control':[10,20,15,20,10,13],
      'treates':[20,30,15,5,13,34]}
```

```python
df=pd.DataFrame(data)
```

```python
df.to_csv("rna_seq.csv",index=False)
```

```python
#calculate folde change
df["folde_change"]=df['treates']/df['control']
```

```python
df
```

```python
status=[]
for x in df["folde_change"]:
    if x>=2:
        status.append('upregulated')
    elif x<=0.5:
        status.append('dounregulated')
    else:
        status.append('no chnage')
```

```python
df['status']=status
```
